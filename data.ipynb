```python
import os
import pandas as pd

# =========================
# Load datasets
# =========================
calibration = pd.read_csv("dataset/calibration.csv")
test_clean = pd.read_csv("dataset/test_clean.csv")
test_noisy = pd.read_csv("dataset/test_noisy.csv")
train = pd.read_csv("dataset/train.csv")

# =========================
# Columns to drop
# =========================
columns_to_drop = [
    # Transient features
    "tr_d01",
    "tr_d12",
    "tr_decay",
    "tr_dc0",
    "tr_dc1",
    "tr_dc2",
    "tr_pk0",
    "tr_vd01",
    "tr_ring",

    # Crest factor / SNR / frequency
    "cf_v",
    "cf_i",
    "snr_v_db",
    "snr_i_db",
    "f_est",




]

# =========================
# Create output folder
# =========================
output_folder = "dataset/processed"
os.makedirs(output_folder, exist_ok=True)

# =========================
# Process and save datasets
# =========================
datasets = {
    "calibration": calibration,
    "test_clean": test_clean,
    "test_noisy": test_noisy,
    "train": train
}

for name, df in datasets.items():
    # Drop specified columns
    df_processed = df.drop(columns=columns_to_drop)

    # Save processed dataset
    output_path = os.path.join(output_folder, f"{name}.csv")
    df_processed.to_csv(output_path, index=False)

    print(f"{name}:")
    print(f"  Original shape : {df.shape}")
    print(f"  New shape      : {df_processed.shape}")
    print(f"  Saved to       : {output_path}")
    print()

print("All datasets processed successfully.")
```
